```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, balanced_accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
import joblib
import shutil
from google.colab import files
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid')

print("="*60)
print("WELLBEING COMPASS - COMPLETE TRAINING PIPELINE")
print("="*60)
print("\nPlease upload your mental_health_dataset.csv file:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]
df_raw = pd.read_csv(filename)

print(f"\nDataset loaded successfully. Shape: {df_raw.shape}")
print("\nFirst 5 rows:")
display(df_raw.head())
print("\nDataset Info:")
df_raw.info()

print("\nMissing values:")
print(df_raw.isnull().sum())
print("\nDuplicate rows:", df_raw.duplicated().sum())

df_raw['Severity'] = df_raw['Severity'].fillna('None')
df_raw['Mental_Health_Condition'] = df_raw['Mental_Health_Condition'].fillna('No')

print("\nTarget distribution (Mental_Health_Condition):")
display(df_raw['Mental_Health_Condition'].value_counts().to_frame('count'))
print("\nSeverity distribution:")
display(df_raw['Severity'].value_counts().to_frame('count'))

numerical_features = ['Age', 'Sleep_Hours', 'Work_Hours', 'Physical_Activity_Hours']
categorical_features = ['Gender', 'Occupation', 'Country', 'Consultation_History', 'Stress_Level']

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
colors = ['#2AA198', '#7559D9', '#F26B6B', '#E9B949']
for ax, column, color in zip(axes.flat, numerical_features, colors):
    sns.histplot(df_raw[column], kde=True, color=color, ax=ax)
    ax.axvline(df_raw[column].mean(), color='#11265C', linestyle='--', label=f'Mean = {df_raw[column].mean():.2f}')
    ax.set_title(column)
    ax.legend()
plt.tight_layout()
plt.show()

display(df_raw[numerical_features].describe().T)

categorical_plot_features = ['Stress_Level', 'Consultation_History', 'Occupation', 'Gender', 'Country']
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
for ax, column in zip(axes.flat, categorical_plot_features):
    order = df_raw[column].value_counts().index
    sns.countplot(data=df_raw, x=column, order=order, color='#2AA198', ax=ax)
    ax.set_title(column)
    ax.tick_params(axis='x', rotation=25)
axes.flat[-1].axis('off')
plt.tight_layout()
plt.show()

stress_target = pd.crosstab(df_raw['Stress_Level'], df_raw['Mental_Health_Condition'], normalize='index') * 100
display(stress_target.round(1))
stress_target[['No', 'Yes']].plot(kind='bar', stacked=True, color=['#2AA198', '#F26B6B'], figsize=(8, 5))
plt.ylabel('Within-group percentage')
plt.title('Mental-health condition rate by stress level')
plt.xticks(rotation=0)
plt.show()

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, column in zip(axes.flat, numerical_features):
    sns.boxplot(data=df_raw, x='Mental_Health_Condition', y=column,
                order=['No', 'Yes'], hue='Mental_Health_Condition',
                palette={'No': '#2AA198', 'Yes': '#F26B6B'}, legend=False, ax=ax)
    ax.set_title(f'{column} by target')
plt.tight_layout()
plt.show()

display(df_raw.groupby('Mental_Health_Condition')[numerical_features].mean().round(2))

sns.pairplot(df_raw[numerical_features + ['Mental_Health_Condition']],
             hue='Mental_Health_Condition',
             palette={'No': '#2AA198', 'Yes': '#F26B6B'},
             corner=True, diag_kind='hist', plot_kws={'alpha': 0.35, 's': 18})
plt.show()

outlier_rows = []
for column in numerical_features:
    q1 = df_raw[column].quantile(0.25)
    q3 = df_raw[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df_raw[column] < lower) | (df_raw[column] > upper)).sum()
    outlier_rows.append([column, q1, q3, lower, upper, count])
outlier_summary = pd.DataFrame(outlier_rows,
                                columns=['Feature', 'Q1', 'Q3', 'Lower', 'Upper', 'Outliers'])
display(outlier_summary.round(2))

df = df_raw.copy()
df['Sleep_Deficit'] = np.maximum(0, 7 - df['Sleep_Hours'])
df['Overtime_Hours'] = np.maximum(0, df['Work_Hours'] - 40)
df['Work_Sleep_Ratio'] = df['Work_Hours'] / (df['Sleep_Hours'] + 0.01)
df['Low_Activity'] = (df['Physical_Activity_Hours'] < 3).astype(int)
df['Sleep_Score'] = np.where(df['Sleep_Hours'] < 6, 0, np.where(df['Sleep_Hours'] <= 8, 1, 0.5))
df['Activity_Score'] = np.where(df['Physical_Activity_Hours'] >= 5, 1, np.where(df['Physical_Activity_Hours'] >= 3, 0.5, 0))
df['Stress_Score'] = df['Stress_Level'].map({'Low': 0, 'Medium': 1, 'High': 2}).fillna(1)
df['Lifestyle_Risk_Score'] = (df['Sleep_Deficit']/7)*0.3 + (df['Overtime_Hours']/40)*0.3 + (1 - df['Activity_Score'])*0.2 + (df['Stress_Score']/2)*0.2

print("\nEngineered Features (first 5 rows):")
engineered_cols = ['Sleep_Deficit', 'Overtime_Hours', 'Work_Sleep_Ratio',
                   'Low_Activity', 'Sleep_Score', 'Activity_Score',
                   'Stress_Score', 'Lifestyle_Risk_Score']
display(df[engineered_cols].head())

feature_cols = ['Age', 'Sleep_Hours', 'Work_Hours', 'Physical_Activity_Hours',
                'Sleep_Deficit', 'Overtime_Hours', 'Work_Sleep_Ratio',
                'Low_Activity', 'Lifestyle_Risk_Score']
categorical_feature_cols = ['Gender', 'Occupation', 'Country', 'Consultation_History', 'Stress_Level']
all_features = feature_cols + categorical_feature_cols

X = df[all_features].copy()
y_binary = df['Mental_Health_Condition'].map({'No': 0, 'Yes': 1}).fillna(0).astype(int)
severity_map = {'None': 0, 'Low': 1, 'Medium': 2, 'High': 3}
y_severity = df['Severity'].map(severity_map).fillna(0).astype(int)

X_train, X_test, y_train_bin, y_test_bin = train_test_split(X, y_binary, test_size=0.2, random_state=42, stratify=y_binary)
X_train_sev, X_test_sev, y_train_sev, y_test_sev = train_test_split(X, y_severity, test_size=0.2, random_state=42, stratify=y_severity)

print(f"\nTraining set size: {X_train.shape[0]}")
print(f"Test set size: {X_test.shape[0]}")
print("\nTraining target distribution (binary):")
display(y_train_bin.value_counts().rename(index={0:'No',1:'Yes'}))

preprocessor = ColumnTransformer([
    ('num', StandardScaler(), feature_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_feature_cols)
])

preprocessor.fit(X_train)

X_train_processed = preprocessor.transform(X_train)
X_test_processed = preprocessor.transform(X_test)
X_train_sev_processed = preprocessor.transform(X_train_sev)
X_test_sev_processed = preprocessor.transform(X_test_sev)

print("\nTraining Binary Classifier with GridSearchCV...")
param_grid = [
    {'classifier': [RandomForestClassifier(random_state=42)], 'classifier__n_estimators': [50, 100], 'classifier__max_depth': [None, 10], 'classifier__class_weight': ['balanced', None]},
    {'classifier': [GradientBoostingClassifier(random_state=42)], 'classifier__n_estimators': [50, 100], 'classifier__learning_rate': [0.05, 0.1], 'classifier__max_depth': [3, 5]},
    {'classifier': [LogisticRegression(max_iter=1000, random_state=42)], 'classifier__C': [0.1, 1.0, 10.0], 'classifier__class_weight': ['balanced', None]},
    {'classifier': [SVC(kernel='rbf', random_state=42, probability=True)], 'classifier__C': [0.1, 1.0, 10.0], 'classifier__gamma': ['scale', 'auto'], 'classifier__class_weight': ['balanced', None]}
]

pipeline = ImbPipeline([
    ('smote', SMOTE(random_state=42, k_neighbors=3)),
    ('classifier', RandomForestClassifier(random_state=42))
])

grid = GridSearchCV(pipeline, param_grid, cv=3, scoring='balanced_accuracy', n_jobs=-1, verbose=1)
grid.fit(X_train_processed, y_train_bin)

best_model = grid.best_estimator_
print(f"\nBest Model: {grid.best_params_}")
y_pred_bin = best_model.predict(X_test_processed)

print("\n" + "="*60)
print("BINARY CLASSIFICATION RESULTS - BEST MODEL")
print("="*60)
print(f"Accuracy: {accuracy_score(y_test_bin, y_pred_bin):.4f}")
print(f"Balanced Accuracy: {balanced_accuracy_score(y_test_bin, y_pred_bin):.4f}")
print(f"Precision: {precision_score(y_test_bin, y_pred_bin):.4f}")
print(f"Recall: {recall_score(y_test_bin, y_pred_bin):.4f}")
print(f"F1-Score: {f1_score(y_test_bin, y_pred_bin):.4f}")
print(f"ROC-AUC: {roc_auc_score(y_test_bin, y_pred_bin):.4f}")

fig, ax = plt.subplots(figsize=(6,5))
ConfusionMatrixDisplay.from_predictions(y_test_bin, y_pred_bin,
                                        display_labels=['No', 'Yes'],
                                        cmap='GnBu', ax=ax)
ax.set_title('Confusion Matrix - Binary Classification (Best Model)')
plt.show()

print("\nTraining Severity Classifier...")
severity_model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
severity_model.fit(X_train_sev_processed, y_train_sev)
y_pred_sev = severity_model.predict(X_test_sev_processed)

print("\n" + "="*60)
print("SEVERITY CLASSIFICATION RESULTS (Multi-Class)")
print("="*60)
print(f"Accuracy: {accuracy_score(y_test_sev, y_pred_sev):.4f}")
print(f"Balanced Accuracy: {balanced_accuracy_score(y_test_sev, y_pred_sev):.4f}")

print("\n" + "="*80)
print("COMPLETE MODEL COMPARISON TABLE (All Models)")
print("="*80)

models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced'),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42, n_estimators=100, learning_rate=0.1, max_depth=3),
    'SVM (RBF)': SVC(kernel='rbf', random_state=42, probability=True, class_weight='balanced', C=1.0, gamma='scale'),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100, max_depth=None, class_weight='balanced')
}

results = []
for name, model in models.items():
    model.fit(X_train_processed, y_train_bin)
    y_pred = model.predict(X_test_processed)

    acc = accuracy_score(y_test_bin, y_pred)
    bacc = balanced_accuracy_score(y_test_bin, y_pred)
    prec = precision_score(y_test_bin, y_pred)
    rec = recall_score(y_test_bin, y_pred)
    f1 = f1_score(y_test_bin, y_pred)

    results.append([name, acc, bacc, prec, rec, f1])
    print(f"{name:25s} | Acc={acc:.4f} | BACC={bacc:.4f} | Prec={prec:.4f} | Rec={rec:.4f} | F1={f1:.4f}")

results_df = pd.DataFrame(results, columns=['Model', 'Accuracy', 'Balanced Accuracy', 'Precision', 'Recall', 'F1-Score'])
print("\n" + "="*80)
print("FORMATTED TABLE")
print("="*80)
display(results_df.round(4))

fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(results_df['Model']))
width = 0.15
metrics = ['Accuracy', 'Balanced Accuracy', 'Precision', 'Recall', 'F1-Score']
colors = ['#2AA198', '#7559D9', '#F26B6B', '#E9B949', '#11265C']

for i, (metric, color) in enumerate(zip(metrics, colors)):
    ax.bar(x + i*width, results_df[metric], width, label=metric, color=color, alpha=0.8)

ax.set_xlabel('Models')
ax.set_ylabel('Score')
ax.set_title('Model Performance Comparison')
ax.set_xticks(x + width*2)
ax.set_xticklabels(results_df['Model'], rotation=15, ha='right')
ax.legend(loc='lower right')
ax.set_ylim(0, 1)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

scaler = preprocessor.named_transformers_['num']
encoder = preprocessor.named_transformers_['cat']

scaler_mean = scaler.mean_
scaler_scale = scaler.scale_
encoder_categories = encoder.categories_

print("\nSaving artifacts (without pipeline) for web deployment...")
export_dir = Path('./deployment_artifacts')
shutil.rmtree(export_dir, ignore_errors=True)
export_dir.mkdir(parents=True, exist_ok=True)

joblib.dump(best_model, export_dir / 'model_binary.pkl')
joblib.dump(severity_model, export_dir / 'model_severity.pkl')
joblib.dump(all_features, export_dir / 'features.pkl')
joblib.dump(severity_map, export_dir / 'severity_map.pkl')
joblib.dump(scaler_mean, export_dir / 'scaler_mean.pkl')
joblib.dump(scaler_scale, export_dir / 'scaler_scale.pkl')
joblib.dump(encoder_categories, export_dir / 'encoder_categories.pkl')
joblib.dump(feature_cols, export_dir / 'feature_cols.pkl')
joblib.dump(categorical_feature_cols, export_dir / 'categorical_feature_cols.pkl')

shutil.make_archive('deployment_artifacts', 'zip', export_dir)
print("\n✅ deployment_artifacts.zip created successfully!")
files.download('deployment_artifacts.zip')
print("\n" + "="*60)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("="*60)
```
